##### 7. Immutable Data Storage

```python
accounts = ('123456789', '987654321', '456789123')
accounts[0] = '111111111'
print(accounts)
```

- 'tuple' object does not support item assignment
- Security , lists are mutable, any part of a large software system with access to that list can modify its contents. Tuples protect the data from unauthorized side
