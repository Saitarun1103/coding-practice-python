##### 19. Consecutive Duplicate Counter (String)

```python
s19 = "aaabbcaaa"
consecutive_count = 1
for i in range(1, len(s19) + 1):
    # Check if we are still within bounds and the current character matches the previous
    if i < len(s19) and s19[i] == s19[i - 1]:
        consecutive_count += 1
    else:
        print(f"{s19[i - 1]} → {consecutive_count}")
        consecutive_count = 1
print("---")
```

##### 20. Anagram Checker 


```python
str1 = input()
str2 = input()

if len(str1) != len(str2):
    print("Not Anagram")
else:
    list2 = list(str2)
    is_anagram = True

    for char in str1:
        if char in list2:
            list2.remove(char)
        else:
            is_anagram = False
            break
    if is_anagram:
        print("Anagram")
    else:
        print("Not Anagram")
```
