##### 17.List Modification Methods

```python
#1. Use append() to add "orange" to the list fruits = ["apple", "banana"].
fruits = ["apple", "banana"]
fruits.append("orange")
print(fruits)

#2. Use insert() to add "grapes" at index 1 in the list fruits = ["apple", "banana"].
fruits = ["apple", "banana"]
fruits.insert(1, "grapes")
print(fruits)

#3. Use extend() to add ["kiwi", "mango"] to the end of fruits = ["apple", "banana"].
fruits = ["apple", "banana"]
fruits.extend(["kiwi", "mango"])
print(fruits)
```

```python
#1 Sort the list marks = [60, 90, 70] in ascending order using sort().
marks = [60, 90, 70]
marks.sort()
print(marks)

#2 Sort the list marks = [60, 90, 70] in descending order using sort(reverse=True).
marks = [60, 90, 70]
marks.sort(reverse=True)
print(marks)

#3 Use reverse() to reverse the list ["a", "b", "c"].
a = ["a", "b", "c"]
a.reverse()
print(a)

#4 What is the difference between sort() and reverse()?
#sort checks the value and arranges, but reverse() just reverse the string.
```
